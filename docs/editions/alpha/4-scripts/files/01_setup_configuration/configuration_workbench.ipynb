# Setup & Configuration Workbench

Loads Capybara configuration, builds the selected ITSM connector, and previews canonical records. Defaults are safe for the included MockConnector.

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
while ROOT.name != "script_files" and ROOT.parent != ROOT:
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
from capybara.core.config import load_config
from capybara.connectors.registry import build_connector
cfg = load_config(ROOT.parent / "config.toml")
connector = build_connector(cfg)
connector.health_check()
```

```python
tickets = list(connector.iter_tickets(limit=5))
assets = list(connector.iter_assets(limit=5))
tickets[:2], assets[:2]
```
