# Alert Preview Workbench

Shows which records would trigger missing-asset, stale-unassigned, and parts/procurement-hold alerts before any SMTP delivery is enabled.

```python
from pathlib import Path
import sys
ROOT=Path.cwd()
while ROOT.name!="script_files" and ROOT.parent!=ROOT: ROOT=ROOT.parent
sys.path.insert(0,str(ROOT))
from capybara.core.config import load_config
from capybara.connectors.registry import build_connector
from capybara.services.notifications import missing_assets, stale_unassigned_tickets, parts_hold_tickets
cfg=load_config(ROOT.parent/"config.toml"); c=build_connector(cfg)
tickets=list(c.iter_tickets()); assets=list(c.iter_assets())
```

```python
missing_assets(assets)
```

```python
stale_unassigned_tickets(tickets,hours=8)
```

```python
parts_hold_tickets(tickets,days=5)
```
