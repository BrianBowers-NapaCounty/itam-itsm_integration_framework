# Reporting Workbench

Builds several management/operations tables from the same canonical data, demonstrating how reports stay connector-agnostic.

```python
from pathlib import Path
import sys, pandas as pd
ROOT=Path.cwd()
while ROOT.name!="script_files" and ROOT.parent!=ROOT: ROOT=ROOT.parent
sys.path.insert(0,str(ROOT))
from capybara.core.config import load_config
from capybara.connectors.registry import build_connector
from capybara.services.reporting import assignment_backlog, tickets_on_hold, service_hotspots, executive_summary
cfg=load_config(ROOT.parent/"config.toml"); c=build_connector(cfg)
tickets=list(c.iter_tickets()); assets=list(c.iter_assets())
```

```python
pd.DataFrame(assignment_backlog(tickets))
```

```python
pd.DataFrame(tickets_on_hold(tickets))
```

```python
pd.DataFrame(service_hotspots(tickets))
```

```python
pd.DataFrame(executive_summary(tickets,assets))
```
