# Spatial Service Explorer

A starting point for joining ticket demand to ArcGIS Indoors Units and exploring building/floor hotspots.

```python
from pathlib import Path
import sys, pandas as pd
ROOT=Path.cwd()
while ROOT.name!="script_files" and ROOT.parent!=ROOT: ROOT=ROOT.parent
sys.path.insert(0,str(ROOT))
from capybara.core.config import load_config
from capybara.connectors.registry import build_connector
from capybara.services.reporting import service_hotspots
cfg=load_config(ROOT.parent/"config.toml"); c=build_connector(cfg)
hotspots=pd.DataFrame(service_hotspots(list(c.iter_tickets())))
hotspots
```
