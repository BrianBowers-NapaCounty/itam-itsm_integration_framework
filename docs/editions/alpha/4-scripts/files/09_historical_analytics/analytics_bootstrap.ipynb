# Historical Analytics Bootstrap

Explores the hosted-table schema, reporting windows, and a local SQLite demo store before any Portal/AGOL writes are enabled.

```python
from pathlib import Path
import sys
ROOT=Path.cwd()
while ROOT.name!="script_files" and ROOT.parent!=ROOT: ROOT=ROOT.parent
sys.path.insert(0,str(ROOT))
from capybara.analytics.schemas import TABLES
[(name,len(defn["fields"])) for name,defn in TABLES.items()]
```

```python
from capybara.analytics.windows import common_windows
for w in common_windows(tz_name="America/Los_Angeles"):
    print(w.code,w.start,w.end,w.label)
```

```python
from capybara.analytics.store import SQLiteAnalyticsStore
store=SQLiteAnalyticsStore(ROOT.parent/"output"/"analytics-demo.sqlite")
store.ensure_tables()
"Local analytics demo store ready"
```
