# Data Quality Scorecard

Calculates a reference governance score from location, assignment, asset-link, known-location, and duplicate-ID quality indicators.

```python
from pathlib import Path
import sys
ROOT=Path.cwd()
while ROOT.name!="script_files" and ROOT.parent!=ROOT: ROOT=ROOT.parent
sys.path.insert(0,str(ROOT))
from capybara.core.config import load_config
from capybara.connectors.registry import build_connector
from capybara.analytics.aggregation import data_quality_row
cfg=load_config(ROOT.parent/"config.toml"); c=build_connector(cfg)
tickets=list(c.iter_tickets())
known={x.strip() for x in (ROOT.parent/"sample_data"/"known_locations.txt").read_text().splitlines() if x.strip()}
score=data_quality_row(tickets,known_locations=known)
score
```

```python
import matplotlib.pyplot as plt
labels=["Location","Assignee","Asset link","Known location"]
vals=[score["LOCATION_COMPLETENESS"],score["ASSIGNEE_COMPLETENESS"],score["ASSET_LINK_RATE"],score["KNOWN_LOCATION_RATE"]]
plt.figure(figsize=(8,3))
plt.bar(labels,vals);plt.ylim(0,1);plt.ylabel("Completeness");plt.xticks(rotation=20);plt.show()
```
