# Monitoring Dashboard

Computes connector health, sync-lag, and location-quality indicators. Optional plotting uses Matplotlib.

```python
from pathlib import Path
import sys
ROOT=Path.cwd()
while ROOT.name!="script_files" and ROOT.parent!=ROOT: ROOT=ROOT.parent
sys.path.insert(0,str(ROOT))
from capybara.core.config import load_config
from capybara.connectors.registry import build_connector
from capybara.services.monitoring import sync_lag_minutes, unmatched_location_rate
cfg=load_config(ROOT.parent/"config.toml")
c=build_connector(cfg)
tickets=list(c.iter_tickets())
known={x.strip() for x in (ROOT.parent/"sample_data"/"known_locations.txt").read_text().splitlines() if x.strip()}
metrics={"connector_ok":c.health_check().ok,
         "sync_lag_minutes":sync_lag_minutes(tickets),
         "unmatched_location_rate":unmatched_location_rate(tickets,known)}
metrics
```

```python
import matplotlib.pyplot as plt
labels=["Location match","Unmatched"]
u=metrics["unmatched_location_rate"]
plt.figure(figsize=(6,3))
plt.bar(labels,[1-u,u])
plt.ylim(0,1)
plt.ylabel("Share of tickets")
plt.show()
```
