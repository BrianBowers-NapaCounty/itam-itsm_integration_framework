# Extension SDK Lab

Shows how local teams can add metrics and enrichment hooks without editing the core connector/reporting packages.

```python
from pathlib import Path
import sys
ROOT=Path.cwd()
while ROOT.name!="script_files" and ROOT.parent!=ROOT: ROOT=ROOT.parent
sys.path.insert(0,str(ROOT))
from capybara.analytics.metrics import metric,REGISTRY
@metric("demo.example","Example local metric","count")
def example_metric(tickets,assets=None,**_):
    return len(tickets)
sorted(REGISTRY)[-5:]
```

```python
from capybara.extensions.hooks import register,emit
@register("enrich_ticket")
def tag(payload,**ctx):
    return {**payload,"demo_tag":"local-extension"}
emit("enrich_ticket",{"ticket":"INC0001"})
```
